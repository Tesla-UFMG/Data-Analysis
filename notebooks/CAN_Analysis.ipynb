```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
import math

plt.rcParams.update({'font.size': 18})
plt.rcParams['figure.figsize'] = (10, 6)

dataset_path = "../data/raw_can_data/28.CSV"
df = pd.read_csv(dataset_path)
df.head(10)
```

```python
ref_timestamp = df['Timestamp'][0]
counter = 0
packets_list = []
for i in range(1, df.shape[0]):
    counter += 1

    if df['Timestamp'][i] - ref_timestamp > 1000000:
        packets_list.append(counter)
        counter = 0
        ref_timestamp = df['Timestamp'][i]
```

```python
plt.rcParams['axes.facecolor'] = '#f0f0f0' # Inner plot area
plt.rcParams['figure.facecolor'] = '#f0f0f0' # Outer area

plt.figure(figsize=(12, 6))
plt.hist(packets_list, bins=100, color='green', alpha=0.7)
plt.xlabel('Number of CAN packets')
plt.ylabel('Frequency')
plt.title('CAN packets per second distribution')
plt.grid(alpha=0.5)
```
